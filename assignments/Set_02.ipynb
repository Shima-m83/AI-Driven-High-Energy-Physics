# Exercise 01: Signal Processing in a PET System

Consider a positron-emission tomography (PET) system that detects two **511 keV** gamma rays produced by positron–electron annihilation.

### 1.1. Positron Source
**Propose a reasonable positron-emission flux (number of positrons emitted per second) from a typical PET radionuclide. Assume the resulting gamma flux enters a liquid scintillator detector.**
### 1.2. Scintillation Process
**A) Discuss at least two different scintillation materials that could be used in such a detector. For each material, describe its approximate light yield and decay time.**

**B) Estimate the total number of scintillation photons produced for a 511 keV gamma-ray interaction and the approximate time width of their emission.**
### 1.3. Detection Chain
Assume the following detector parameters:
- Light collection efficiency: 60%
- Photocathode quantum efficiency: 20%
- Photomultiplier tube (PMT) total gain: $10^{6}$

**A) Calculate the number of electrons reaching the first dynode and the number of electrons at the anode for one event.**

**B) If the signal is read out through a $50 \Omega$ load resistor, compute the approximate voltage pulse amplitude produced by the event.**
### 1.4. Statistical Energy Resolution
**Assuming the number of photoelectrons follows Poisson statistics, determine the fractional energy resolution ($\frac{\sigma_E}{E}$) of the detector output.**
### 1.5. Noise Discussion
**A) Identify and discuss the possible sources of noise in this detection system, such as statistical fluctuations, gain variations, and electronic or thermal noise.**

**B) Compare the statistical fluctuation computed above with the expected thermal noise level in the electronics, and comment on which contribution dominates.**


#### Your answer:

**1.1 Positron source (reasonable flux).**  
Typical clinical PET injections are in the 10–400 MBq range; take a modest activity $A=10\,\text{MBq}=10^{7}\,\text{decays/s}$. Each positron emission (decay) ultimately produces a pair of 511 keV gamma rays. If the detector subtends a small solid angle fraction $f_{\Omega}$ (e.g. $10^{-3}$–$10^{-4}$) the gamma flux entering the detector is roughly

$$\Phi_{\gamma} \approx A \times f_{\Omega} \sim 10^{7} \times 10^{-4} = 10^{3}\text{--}10^{4}\ \text{gamma/s}. $$

**1.2 Scintillation process.**  
A) Two common scintillators:

- **LYSO ($Lu_{2}$ $SiO_5$:Ce)** — often used in PET: light yield ≈ 25–35k photons/MeV, decay time ≈ 40 ns (fast, good timing).  
- **NaI(Tl)** — high light yield ≈ 38k photons/MeV, but slower decay ≈ 230 ns (good energy resolution, poor timing).  
(Also: plastic scintillators: light yield ≈ 8–12k ph/MeV, decay few ns — very fast but low-Z so poor 511 keV stopping power.)

B) For a full-energy deposition of a 511 keV gamma (0.511 MeV), the number of scintillation photons produced is approximately

$N_{\mathrm{ph}} \approx (\text{light yield}) \times 0.511.$ 

Example (LYSO, 30k ph/MeV): $N_{\mathrm{ph}} \approx 30{,}000 \times 0.511 \approx 15{,}330$ photons. The emission is spread over the scintillator decay time (tens to hundreds of ns): the characteristic time width is of order the decay time (e.g. 40 ns for LYSO, 230 ns for NaI(Tl)).

**1.3 Detection chain (using the given parameters).**  
Assume $N_{\mathrm{ph}}$ photons created at the scintillator, light collection $\eta_{\text{coll}}=0.60$, photocathode QE $\eta_d=0.20$.

A) Number of photoelectrons at the photocathode:

$N_{pe} = N_{\mathrm{ph}} \times \eta_{\text{coll}} \times \eta_d.$ 

Using the LYSO example above: $N_{pe} \approx 15{,}330 \times 0.6 \times 0.2 \approx 1{,}840$ photoelectrons. Electrons reaching the first dynode are approximately $N_{pe}$. After PMT gain $G=10^{6}$, total electrons at anode:

$$N_{\mathrm{anode}} \approx N_{pe} \times G \approx 1.84\times10^{3} \times 10^{6} \approx 1.84\times10^{9}\ \text{electrons}. $$

B) Voltage pulse across a $50\,\Omega$ load: total charge $Q = N_{\mathrm{anode}} \times e$ (with $e\approx 1.6\times 10^{-19}\,$ C). If this charge is delivered in a pulse of duration $\tau$ (say $\tau \sim 50\,$ ns typical), the average current $I \approx Q/\tau$ and the pulse amplitude $V=I\times 50\,\Omega$.

Numerical estimate (using $\tau = 50\,$ ns):

$$Q \approx 1.84\times10^{9} \times 1.6\times10^{-19}\ \mathrm{C} \approx 2.94\times10^{-10}\ \mathrm{C},$$
$$I \approx \frac{Q}{\tau} \approx \frac{2.94\times10^{-10}}{5\times10^{-8}} \approx 5.9\times10^{-3}\ \mathrm{A},$$
$$V \approx I \times 50\,\Omega \approx 0.295\ \mathrm{V} \ (\text{~}300\ \mathrm{mV}).$$


**1.4 Statistical energy resolution.**  
If the dominant fluctuation is the Poisson statistics of detected photoelectrons, then

$\frac{\sigma_E}{E} \approx \frac{1}{\sqrt{N_{pe}}}.$ 

With $N_{pe} \approx 1{,}840$ this gives $\sigma_E/E \approx 1/\sqrt{1840} \approx 0.023$ or ≈2.3% (photon-statistics limit; other factors worsen this).

**1.5 Noise discussion.**  
A) Main noise sources: photoelectron (Poisson) statistics, PMT gain fluctuations (excess noise factor), electronic noise (preamplifier thermal noise and readout noise), dark counts (PMT/photodiode) and cross-talk. Timing jitter can also limit time resolution.  

B) Comparison: for the numbers above the photoelectron statistical term (2–5%) is typically dominant for energy resolution if light yield is modest; electronics noise produces a fixed-equivalent noise in electrons and matters most when $N_{pe}$ is small. Gain fluctuations (PMT excess noise factor) increase the effective variance by a factor $F$ (often 1.2–1.5), so include that by replacing $N_{pe}\to N_{pe}/F$ in the resolution estimate. In many PET detectors with good scintillators and light collection, statistical fluctuations at the photocathode and PMT gain are the main limits; for very low-light or poor readout electronics, electronics noise can dominate.


# Exercise 02: CCD Noise Estimation with Finite Background Sampling

**A)** Show for an ideal CCD with measurement of the background using a large number of pixels in the vicinity of the signal, the SNR is given by

$$\mathrm{SNR} = \dfrac{ \sqrt{\eta_d N_{\nu S}}}{\sqrt{1 + n_{\text{pix}} \left(\dfrac{N_{\nu B}}{N_{\nu S}}\right)}}$$

Where $\eta_d$ is responsive quantum efficiency. $N_{\nu S}$ is total number of detected signal photons (summed over all $n_{\text{pix}}$ pixels). $N_{\nu B}$  mean number of background photons per pixel during the exposure.

**B)** Show how the SNR expression is modified when the background level is estimated from a finite number of background pixels, $n_{\text{bck}}$, comparable to $n_{\text{pix}}$.

**C)** In a real CCD, there will be additional noise contributions from dark current and read noise. Suppose $I$ is the dark current per pixel in unit of electrons per unit time. Also Read noise introduces a fixed charge uncertainty, $\Delta q_R$, per pixel, regardless of the integration time. Allowing for these noise contributions show that the read noise contribution to the overall noise decreases with integration time, but the dark current contribution does not.


#### Your answer:

**A) Derivation when background is estimated perfectly (or not subtracted):**  
Let $N_{\nu S}$ be the number of signal photons incident on the source aperture (over all aperture pixels) and $N_{\nu B}$ the mean number of background photons per pixel (both are *incident* photons). The detected photoelectron numbers scale by the detector efficiency $\eta_d$. For an aperture of $n_{\text{pix}}$ pixels, the expected detected signal electrons are $\eta_d N_{\nu S}$ and the expected background electrons inside the aperture are $\eta_d n_{\text{pix}} N_{\nu B}$. Assuming Poisson statistics and no other noise, the variance of the measured (signal + background) counts is

$\mathrm{Var} = \eta_d N_{\nu S} + \eta_d n_{\text{pix}} N_{\nu B}.$ 

If we subtract the (known) background mean, the net-signal mean is $\eta_d N_{\nu S}$ and the noise RMS is $\sqrt{\eta_d\left(N_{\nu S}+n_{\text{pix}}N_{\nu B}\right)}$. Therefore

$$\mathrm{SNR} = \frac{\eta_d N_{\nu S}}{\sqrt{\eta_d\left(N_{\nu S}+n_{\text{pix}}N_{\nu B}\right)}} = \frac{\sqrt{\eta_d N_{\nu S}}}{\sqrt{1 + n_{\text{pix}}\dfrac{N_{\nu B}}{N_{\nu S}}}},$$

**B) Background estimated from $n_{\text{bck}}$ pixels (finite-sample background).**  
When the background level is not known a priori and is estimated from $n_{\text{bck}}$ pixels (e.g. nearby sky pixels), the mean background per pixel has variance $\eta_d N_{\nu B}/n_{\text{bck}}$ (Poisson statistics). When this mean is scaled to the aperture (multiplied by $n_{\text{pix}}$) and subtracted, the subtraction contributes an extra variance term

$\mathrm{Var}_{\text{bck,est}} = n_{\text{pix}}^{2} \frac{\eta_d N_{\nu B}}{n_{\text{bck}}}.$ 

Adding this to the measurement variance gives total variance

$\mathrm{Var}_{\text{tot}} = \eta_d N_{\nu S} + \eta_d n_{\text{pix}} N_{\nu B} + n_{\text{pix}}^{2} \frac{\eta_d N_{\nu B}}{n_{\text{bck}}}.$ 

Hence the SNR becomes

$\displaystyle \mathrm{SNR} = \frac{\sqrt{\eta_d N_{\nu S}}}{\sqrt{1 + n_{\text{pix}}\dfrac{N_{\nu B}}{N_{\nu S}} \; + \; \dfrac{n_{\text{pix}}^{2}}{n_{\text{bck}}}\dfrac{N_{\nu B}}{N_{\nu S}} }}.$ 

You can rearrange the denominator to show the middle and last terms combined.

**C) Additional noise terms in a real CCD.**  
Include read noise $R$ (electrons RMS per pixel) and dark current $D$ (electrons per pixel accumulated during integration). The variance contributions per pixel are $R^{2}$ (read noise) and $D$ (Poisson dark current). Propagating these leads to

$\mathrm{Var}_{\text{tot}} = \eta_d N_{\nu S} + n_{\text{pix}}(\eta_d N_{\nu B} + D + R^{2}) + n_{\text{pix}}^{2} \frac{\eta_d N_{\nu B} + D + R^{2}}{n_{\text{bck}}}.$ 

Dark current $D$ typically grows linearly with integration time, whereas read noise $R$ is independent of integration time. For long exposures dark current can dominate; for short integrations read noise and photon statistics dominate.


# Exercise 03: Optical Properties and Magnification of Telescopes Observing the Moon

The Moon was photographed with a telescope, the objective of which had a diameter of 20 cm and focal length of 150 cm. The exposure time was 0.1 s. 

**A)** What should the exposure time be, if the diameter of the objective were 15 cm and focal length 200 cm?

**B)** What is the size of the image of the Moon in both cases? 

**C)** Both telescopes are used to look at the Moon with an eyepiece the focal length of which is 25 mm. What are the magnifications?

#### Your answer:

Given: Moon angular diameter ≈ 0.5° (≈ 0.00872665 rad).
Telescope 1: objective diameter D1 = 20 cm, focal length f1 = 150 cm, exposure t1 = 0.1 s.
Telescope 2: D2 = 15 cm, f2 = 200 cm. Eyepiece focal length fe = 25 mm = 2.5 cm.

---
A) Exposure time for telescope 2

For imaging an extended object with the same detector and required image illumination, the image irradiance scales as

$$E \propto \frac{D^{2}}{f^{2}},$$
so the exposure time required scales as

$t \propto \frac{f^{2}}{D^{2}}.$ 

Thus

$t_{2} = t_{1} \times \frac{(f_{2}/D_{2})^{2}}{(f_{1}/D_{1})^{2}}.$ 

Plugging numbers:

$t_{2} = 0.1\,\text{s} \times \frac{(200/15)^{2}}{(150/20)^{2}} = 0.3160\,\text{s}.$ 

So $t_{2} \approx 0.3160\,$ s (≈ 0.316 s).

---
B) Size of the image of the Moon

Using the small-angle approximation, linear image diameter on the focal plane is

$$s = f \times \theta,$$
where $\theta$ is the Moon's angular diameter in radians (0.00872665 rad).

For telescope 1 (f1 = 150 cm):

$s_{1} = 150\,\text{cm} \times 0.00872665 = 1.3090\,\text{cm}.$ 

For telescope 2 (f2 = 200 cm):

$s_{2} = 200\,\text{cm} \times 0.00872665 = 1.7453\,\text{cm}.$ 

So image sizes: telescope 1 → 1.309 cm, telescope 2 → 1.745 cm.

---
C) Magnifications using an eyepiece with focal length 25 mm (2.5 cm)

Angular magnification of a simple telescope (objective + eyepiece) is

$M = \frac{f_{\text{objective}}}{f_{\text{eyepiece}}}.$ 

Thus

$M_{1} = \frac{150}{2.5} = 60\times, \qquad M_{2} = \frac{200}{2.5} = 80\times.$ 

So magnifications: telescope 1 → 60×, telescope 2 → 80×.

---

Summary (numeric):

- Exposure time for telescope 2: 0.3160 s.
- Moon image diameter on focal plane: Telescope 1: 1.309 cm, Telescope 2: 1.745 cm.
- Angular magnifications with a 25 mm eyepiece: Telescope 1: 60×, Telescope 2: 80×.


# Exercise 04: Maximum Likelihood Estimator

Consider n data points draw independently from a gaussian distribution. Starting from maximum likelihood estimator, find that it equivalent to minimizing the sum-of-squares error function.


#### Your answer:

Assume $x_{1},\dots,x_{n}$ are independent draws from $\mathcal{N}(\mu,\sigma^{2})$. The likelihood is

$$L(\mu,\sigma^{2}) = \prod_{i=1}^{n} \frac{1}{\sqrt{2\pi\sigma^{2}}} \exp\left(-\frac{(x_{i}-\mu)^{2}}{2\sigma^{2}}\right).$$

Log-likelihood:

$\ell(\mu,\sigma^{2}) = -\frac{n}{2}\ln(2\pi\sigma^{2}) - \frac{1}{2\sigma^{2}} \sum_{i=1}^{n} (x_{i}-\mu)^{2}.$ 

Maximizing $\ell$ w.r.t. $\mu$ (fixing $\sigma^{2}$) is equivalent to minimizing $\sum_{i}(x_{i}-\mu)^{2}$. Differentiate:

$$\frac{\partial\ell}{\partial\mu} = -\frac{1}{\sigma^{2}} \sum_{i} (x_{i}-\mu) = 0 \quad\Rightarrow\quad \hat{\mu} = \frac{1}{n} \sum_{i} x_{i},$$

the sample mean.

Maximizing w.r.t. $\sigma^{2}$, set $\partial\ell/\partial\sigma^{2}=0$ to obtain the MLE

$$\widehat{\sigma}^{2}_{\mathrm{mle}} = \frac{1}{n} \sum_{i=1}^{n} (x_{i}-\hat{\mu})^{2},$$

(i.e. the MLE uses $1/n$, not $1/(n-1)$). Thus the MLE for $\mu$ is the sample mean and maximizing the likelihood is equivalent to minimizing the sum-of-squares error function.
